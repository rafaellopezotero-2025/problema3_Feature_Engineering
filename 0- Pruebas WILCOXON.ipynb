#** PRUEBAS WILCOXON **

```python
##PRUEBA EXP3010 vs EXP3011: baseline vs ratios de 10 variables mas importantes

import numpy as np
from scipy.stats import wilcoxon

# Resultados de las 10 semillas
exp3010 = np.array([
    34.5475,
    33.0775,
    32.9100,
    33.0225,
    33.6250,
    33.8850,
    31.7725,
    33.9150,
    32.5850,
    33.6775
])

exp3011 = np.array([
    31.6475,
    30.9550,
    33.3425,
    32.0275,
    33.3625,
    34.1600,
    32.2100,
    33.5225,
    32.7700,
    32.7700
])

# Wilcoxon pareado
resultado = wilcoxon(
    exp3010,
    exp3011,
    alternative="two-sided"
)

# Resultados
print("Wilcoxon pareado")
print("-----------------")
print(f"Mediana exp3010: {np.median(exp3010):.3f}")
print(f"Mediana exp3011: {np.median(exp3011):.3f}")
print(f"Estadístico W: {resultado.statistic:.3f}")
print(f"p-value: {resultado.pvalue:.6f}")

if resultado.pvalue < 0.05:
    print("Hay diferencias significativas entre los tratamientos (p < 0.05).")
else:
    print("No hay diferencias significativas entre los tratamientos (p >= 0.05).")
```
